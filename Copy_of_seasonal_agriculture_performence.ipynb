```python
# Data Analysis Libraries
import pandas as pd
import numpy as np

# Visualization Libraries
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

print("Libraries imported successfully!")
```

```python
# Load the dataset

file_name = "seasonal_agriculture_performance_dataset.csv"

df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python
# Dataset information

print("Dataset Information:")
df.info()
```

```python
# Check missing values and duplicate records

print("Missing Values:")
display(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Statistical summary of numerical columns

display(df.describe().T)
```

```python
# DATA CLEANING
# Create a copy of original dataset
df_clean = df.copy()

# Remove duplicate rows
duplicates_before = df_clean.duplicated().sum()
df_clean = df_clean.drop_duplicates()

# Numerical columns
numeric_cols = df_clean.select_dtypes(include=np.number).columns

# Fill missing numerical values with median
for col in numeric_cols:
    if df_clean[col].isnull().sum() > 0:
        df_clean[col] = df_clean[col].fillna(df_clean[col].median())

# Fill missing categorical values with mode
categorical_cols = df_clean.select_dtypes(include='object').columns

for col in categorical_cols:
    if df_clean[col].isnull().sum() > 0:
        df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

print("Data cleaning completed successfully!")
print("Duplicates removed:", duplicates_before)
print("Remaining missing values:", df_clean.isnull().sum().sum())
print("Final dataset shape:", df_clean.shape)
```

```python
# Check number of records in each season

season_count = df_clean['Season'].value_counts()

print("Number of records by season:")
display(season_count)
```

```python

# GRAPH 1: AVERAGE YIELD BY SEASON
season_yield = (
    df_clean.groupby('Season')['Yield_Tonnes_Ha']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    x=season_yield.index,
    y=season_yield.values
)

plt.title(
    'Average Agricultural Yield by Season',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Season', fontsize=12)
plt.ylabel('Average Yield (Tonnes/Ha)', fontsize=12)

# Add values on bars
for i, value in enumerate(season_yield.values):
    ax.text(
        i,
        value,
        f'{value:.2f}',
        ha='center',
        va='bottom',
        fontsize=11,
        fontweight='bold'
    )

plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# GRAPH 2: AVERAGE PROFIT BY SEASON
season_profit = (
    df_clean.groupby('Season')['Profit_INR']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    x=season_profit.index,
    y=season_profit.values
)

plt.title(
    'Average Profit by Agricultural Season',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Season', fontsize=12)
plt.ylabel('Average Profit (INR)', fontsize=12)

for i, value in enumerate(season_profit.values):
    ax.text(
        i,
        value,
        f'₹{value:,.0f}',
        ha='center',
        va='bottom',
        fontsize=10,
        fontweight='bold'
    )

plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# GRAPH 3: CROP-WISE AVERAGE PRODUCTION

crop_production = (
    df_clean.groupby('Crop')['Production_Tonnes']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    x=crop_production.index,
    y=crop_production.values
)

plt.title(
    'Average Production by Crop',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Crop', fontsize=12)
plt.ylabel('Average Production (Tonnes)', fontsize=12)

plt.xticks(rotation=30)

for i, value in enumerate(crop_production.values):
    ax.text(
        i,
        value,
        f'{value:,.1f}',
        ha='center',
        va='bottom',
        fontsize=9
    )

plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# GRAPH 4: RAINFALL VS YIELD
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_clean,
    x='Rainfall_mm',
    y='Yield_Tonnes_Ha',
    alpha=0.6
)

# Trend line
sns.regplot(
    data=df_clean,
    x='Rainfall_mm',
    y='Yield_Tonnes_Ha',
    scatter=False
)

plt.title(
    'Relationship Between Rainfall and Agricultural Yield',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Rainfall (mm)', fontsize=12)
plt.ylabel('Yield (Tonnes/Ha)', fontsize=12)

plt.grid(True, linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# GRAPH 5: PRODUCTION BY SEASON

season_production = (
    df_clean.groupby('Season')['Production_Tonnes']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    x=season_production.index,
    y=season_production.values
)

plt.title(
    'Average Agricultural Production by Season',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Season', fontsize=12)
plt.ylabel('Average Production (Tonnes)', fontsize=12)

for i, value in enumerate(season_production.values):
    ax.text(
        i, value,
        f'{value:,.1f}',
        ha='center',
        va='bottom',
        fontweight='bold'
    )

plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# GRAPH 6: WATER USAGE VS YIELD
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_clean,
    x='Water_Used_m3',
    y='Yield_Tonnes_Ha',
    hue='Season',
    alpha=0.65
)

plt.title(
    'Relationship Between Water Usage and Yield',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Water Used (m³)', fontsize=12)
plt.ylabel('Yield (Tonnes/Ha)', fontsize=12)

plt.grid(True, linestyle='--', alpha=0.3)
plt.legend(title='Season')
plt.tight_layout()
plt.show()
```

```python
# GRAPH 7: TEMPERATURE VS YIELD

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_clean,
    x='Avg_Temperature_C',
    y='Yield_Tonnes_Ha',
    hue='Season',
    alpha=0.65
)

plt.title(
    'Relationship Between Temperature and Agricultural Yield',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Average Temperature (°C)', fontsize=12)
plt.ylabel('Yield (Tonnes/Ha)', fontsize=12)

plt.grid(True, linestyle='--', alpha=0.3)
plt.legend(title='Season')
plt.tight_layout()
plt.show()
```

```python
# GRAPH 8: CORRELATION HEATMAP

plt.figure(figsize=(16, 10))

correlation = df_clean.select_dtypes(
    include=np.number
).corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0,
    linewidths=0.5
)

plt.title(
    'Correlation Heatmap of Agricultural Variables',
    fontsize=18,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```

```python
# GRAPH 9: SEASON × CROP YIELD

season_crop_yield = pd.pivot_table(
    df_clean,
    values='Yield_Tonnes_Ha',
    index='Crop',
    columns='Season',
    aggfunc='mean'
)

plt.figure(figsize=(12, 7))

sns.heatmap(
    season_crop_yield,
    annot=True,
    fmt='.2f',
    cmap='YlGn',
    linewidths=0.5
)

plt.title(
    'Average Crop Yield Across Different Seasons',
    fontsize=17,
    fontweight='bold'
)

plt.xlabel('Season', fontsize=12)
plt.ylabel('Crop', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# GRAPH 10: TOP PERFORMING CROPS

top_crops = (
    df_clean.groupby('Crop')['Yield_Tonnes_Ha']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    x=top_crops.values,
    y=top_crops.index
)

plt.title(
    'Top Performing Crops by Average Yield',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Average Yield (Tonnes/Ha)', fontsize=12)
plt.ylabel('Crop', fontsize=12)

for i, value in enumerate(top_crops.values):
    ax.text(
        value,
        i,
        f' {value:.2f}',
        va='center',
        fontweight='bold'
    )

plt.grid(axis='x', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# SEASONAL PERFORMANCE SUMMARY

season_summary = df_clean.groupby('Season').agg(
    Average_Yield=('Yield_Tonnes_Ha', 'mean'),
    Average_Production=('Production_Tonnes', 'mean'),
    Average_Revenue=('Revenue_INR', 'mean'),
    Average_Profit=('Profit_INR', 'mean'),
    Average_Water_Used=('Water_Used_m3', 'mean'),
    Average_Water_Efficiency=('Water_Efficiency_t_per_1000m3', 'mean')
).round(2)

display(season_summary)
```

```python
# BEST PERFORMING SEASON

best_yield_season = season_summary['Average_Yield'].idxmax()
best_profit_season = season_summary['Average_Profit'].idxmax()
best_production_season = season_summary['Average_Production'].idxmax()
best_water_efficiency_season = season_summary['Average_Water_Efficiency'].idxmax()

print("Best Season by Average Yield:", best_yield_season)
print("Best Season by Average Production:", best_production_season)
print("Best Season by Average Profit:", best_profit_season)
print("Best Season by Water Efficiency:", best_water_efficiency_season)
```

```python
# CROP PERFORMANCE SUMMARY


crop_summary = df_clean.groupby('Crop').agg(
    Average_Yield=('Yield_Tonnes_Ha', 'mean'),
    Average_Production=('Production_Tonnes', 'mean'),
    Average_Profit=('Profit_INR', 'mean')
).sort_values(
    'Average_Yield',
    ascending=False
).round(2)

display(crop_summary)
```

```python
# VARIABLES MOST RELATED TO YIELD

numeric_data = df_clean.select_dtypes(include=np.number)

yield_corr = (
    numeric_data.corr()['Yield_Tonnes_Ha']
    .drop('Yield_Tonnes_Ha')
    .sort_values(key=abs, ascending=False)
)

display(yield_corr.to_frame('Correlation_with_Yield'))
```

```python
# FINAL GRAPH: SEASONAL PERFORMANCE

fig, ax1 = plt.subplots(figsize=(11, 6))

seasons = season_summary.index

yield_values = season_summary['Average_Yield']
profit_values = season_summary['Average_Profit']

# Yield
bars = ax1.bar(
    seasons,
    yield_values,
    alpha=0.8,
    label='Average Yield'
)

ax1.set_xlabel('Season', fontsize=12)
ax1.set_ylabel('Average Yield (Tonnes/Ha)', fontsize=12)

ax1.set_title(
    'Seasonal Agricultural Performance',
    fontsize=17,
    fontweight='bold'
)

# Values on bars
for bar, value in zip(bars, yield_values):
    ax1.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f'{value:.2f}',
        ha='center',
        va='bottom',
        fontweight='bold'
    )

ax1.grid(axis='y', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```
